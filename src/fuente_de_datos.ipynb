## Descarga del dataset

Si el dataset no está disponible localmente, se descarga desde Kaggle y se deposita en `dataset/`. Las rutas son relativas a la raíz del proyecto.

```python
DATASET_DIR = Path.cwd().parent / "dataset"
LOCAL_FILE = DATASET_DIR / "sales_transaction.csv"

if not LOCAL_FILE.exists():
    cached = Path(kagglehub.dataset_download("gabrielramos87/an-online-shop-business"))
    DATASET_DIR.mkdir(parents=True, exist_ok=True)
    LOCAL_FILE.write_bytes((cached / "Sales Transaction v.4a.csv").read_bytes())
    print(f"Dataset descargado en: {LOCAL_FILE}")
else:
    print(f"Dataset ya disponible: {LOCAL_FILE}")
```

## Lectura y limpieza inicial

El archivo CSV se lee con latin1 y la columna de fecha se convierte a datetime para habilitar el análisis temporal.

```python
df = pd.read_csv(LOCAL_FILE, encoding="latin1")
df["Date"] = pd.to_datetime(df["Date"], format="mixed")
print(f"Dimensiones: {df.shape}")
```

## Tabla resumen de la estructura

Se generan las métricas que alimentan la tabla de la subsección Estructura del informe.

```python
print(f"Filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")
print(f"Fecha inicial: {df['Date'].min().date()}")
print(f"Fecha final: {df['Date'].max().date()}")
print(f"Países: {df['Country'].nunique()}")
print(f"Duplicados: {df.duplicated().sum()}")
print(f"Nulos en CustomerNo: {df['CustomerNo'].isna().sum()}")
```

## Diccionario de datos

Se listan las variables con su tipo nativo, su naturaleza y el número de nulos, tal como aparece en el diccionario del informe.

```python
naturaleza = {
    "TransactionNo": "Categórico",
    "Date": "Fecha",
    "ProductNo": "Categórico",
    "ProductName": "Categórico",
    "Price": "Numérico",
    "Quantity": "Numérico",
    "CustomerNo": "Categórico",
    "Country": "Categórico",
}
datos = []
for col in df.columns:
    datos.append({
        "Variable": col,
        "Tipo Dato": str(df[col].dtype).replace("datetime64[us]", "datetime"),
        "Tipo": naturaleza[col],
        "Nulos": int(df[col].isna().sum()),
    })
pd.DataFrame(datos)
```
